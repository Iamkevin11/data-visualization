```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Set a general plotting style
sns.set_theme(style="whitegrid")

# LOAD THE TAXIS DATASET


df = sns.load_dataset("taxis")

# Display the first five rows
print("First five rows:")
print(df.head())

# Display the number of rows and columns
print("\nDataset shape:")
print(df.shape)

# Display column names and data types
print("\nDataset information:")
df.info()

# Display statistical summary
print("\nStatistical summary:")
print(df.describe(include="all"))


# HANDLING MISSING VALUES


# Check the number of missing values in each column
print("\nMissing values before cleaning:")
print(df.isnull().sum())

# Display only columns that contain missing values
print("\nColumns containing missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])


# Convert pickup and dropoff columns into datetime format
df["pickup"] = pd.to_datetime(df["pickup"], errors="coerce")
df["dropoff"] = pd.to_datetime(df["dropoff"], errors="coerce")

# Remove rows where pickup time or fare is missing.
# These columns are important for the required visualizations.
df = df.dropna(subset=["pickup", "fare"]).copy()

# Identify numerical and categorical columns
numerical_columns = df.select_dtypes(include="number").columns
categorical_columns = df.select_dtypes(
    include=["object", "category", "string"]
).columns

# Fill missing numerical values with the median
for column in numerical_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].median())

# Fill missing categorical values with the mode
for column in categorical_columns:
    if df[column].isnull().any():
        mode_values = df[column].mode(dropna=True)

        if not mode_values.empty:
            df[column] = df[column].fillna(mode_values.iloc[0])

# Check missing values after cleaning
print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDataset shape after cleaning:")
print(df.shape)



# LINE CHART USING MATPLOTLIB


# Sort data by pickup time
line_data = df.sort_values("pickup").copy()

# Group by pickup timestamp to make the line chart easier to read
line_data = line_data.set_index("pickup")["fare"].resample("D").mean()

plt.figure(figsize=(12, 6))
plt.plot(line_data.index, line_data.values, marker="o", markersize=3)

plt.title("Average Taxi Fare Over Time")
plt.xlabel("Pickup Date")
plt.ylabel("Average Fare")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



# BAR CHART USING PANDAS


borough_fare = df.groupby(
    "pickup_borough", observed=True
)["fare"].sum().sort_values(ascending=False)

borough_fare.plot(
    kind="bar",
    figsize=(9, 5),
    title="Total Fare by Pickup Borough",
    xlabel="Pickup Borough",
    ylabel="Total Fare"
)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# PIE CHART USING MATPLOTLIB

payment_counts = df["payment"].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(
    payment_counts,
    labels=payment_counts.index.astype(str),
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Trips by Payment Method")
plt.tight_layout()
plt.show()


# HISTOGRAM USING MATPLOTLIB


plt.figure(figsize=(9, 5))
plt.hist(
    df["distance"].dropna(),
    bins=30,
    edgecolor="black"
)

plt.title("Distribution of Taxi Trip Distance")
plt.xlabel("Distance")
plt.ylabel("Number of Trips")
plt.tight_layout()
plt.show()



# BOX PLOT USING MATPLOTLIB


box_data = df.dropna(subset=["pickup_borough", "tip"])

boroughs = box_data["pickup_borough"].dropna().unique()

tip_values = [
    box_data.loc[
        box_data["pickup_borough"] == borough, "tip"
    ].values
    for borough in boroughs
]

plt.figure(figsize=(10, 6))
plt.boxplot(tip_values, tick_labels=boroughs)

plt.title("Distribution of Tips by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Tip Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



# COUNT PLOT USING SEABORN

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="pickup_borough",
    order=df["pickup_borough"].value_counts().index
)

plt.title("Number of Trips by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Number of Trips")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



# SCATTER PLOT USING SEABORN


plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="distance",
    y="fare",
    hue="pickup_borough",
    alpha=0.6
)

plt.title("Relationship Between Distance and Fare")
plt.xlabel("Distance")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()


# HEATMAP USING SEABORN


correlation_columns = [
    "distance", "fare", "tip", "tolls", "total"
]

correlation_matrix = df[correlation_columns].corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Taxi Variables")
plt.tight_layout()
plt.show()



#  PAIR PLOT USING SEABORN


pair_columns = ["distance", "fare", "tip", "total", "pickup_zone"]

# Use the five most common pickup zones so the plot remains readable
top_zones = df["pickup_zone"].value_counts().head(5).index

pair_data = df.loc[
    df["pickup_zone"].isin(top_zones),
    pair_columns
].dropna()

# Limit the number of rows to keep plotting efficient
if len(pair_data) > 500:
    pair_data = pair_data.sample(n=500, random_state=42)

sns.pairplot(
    data=pair_data,
    vars=["distance", "fare", "tip", "total"],
    hue="pickup_zone",
    diag_kind="hist",
    corner=True
)

plt.suptitle(
    "Pairwise Relationships by Pickup Zone",
    y=1.02
)

plt.show()



# VIOLIN PLOT USING SEABORN


violin_data = df.dropna(subset=["payment", "fare"])

plt.figure(figsize=(9, 6))

sns.violinplot(
    data=violin_data,
    x="payment",
    y="fare",
    inner="quartile",
    cut=0
)

plt.title("Fare Distribution by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()


#  FINAL SUMMARY


print("\nAssignment completed successfully!")
print("Total rows after cleaning:", len(df))
print("Total columns:", len(df.columns))
print("Missing values remaining in the dataset:", df.isnull().sum().sum())
```
